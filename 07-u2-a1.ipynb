```python
linguagens = ["Python", "Java", "Javascript", "C#"] # LISTA
```

```python
print("Antes de listcomp =", linguagens)
```

```python
linguagens = [item.lower() for item in linguagens]
print(f"\nDepois de listcomp=", linguagens)
```

```python
convidados = ("Alice","Rodrigo","Karoliny","Aluno")
confirmados = ["Alice","Rodrigo"]
```

```python
print("Convidados que ainda não confirmaram:")
for convidado in convidados:
  if convidado not in confirmados:
    print(f"- {convidado}")
```
