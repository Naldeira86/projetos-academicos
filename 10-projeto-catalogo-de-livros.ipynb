```python
# SISTEMA DE CATALOGAÇÃO DE LIVROS

import matplotlib.pyplot as plt


# Classe que representa um livro
class Livro:

    # Construtor da classe
    def __init__(self, titulo, autor, ano_publicacao):
        self.titulo = titulo
        self.autor = autor
        self.ano_publicacao = ano_publicacao

    # Define como o livro será exibido
    def __str__(self):
        return (
            f"{self.titulo} por {self.autor}, "
            f"Publicado em {self.ano_publicacao}"
        )


# Lista que armazena os livros
biblioteca = []


# Adiciona um livro à biblioteca
def adicionar_livro(titulo, autor, ano_publicacao):

    novo_livro = Livro(titulo, autor, ano_publicacao)
    biblioteca.append(novo_livro)

    print(f"O livro '{titulo}' foi adicionado à biblioteca.")


# Lista os livros cadastrados
def listar_livros():

    print("\n===== LIVROS NA BIBLIOTECA =====")

    for livro in biblioteca:
        print(livro)


# Cadastro dos livros
adicionar_livro("Dom Quixote", "Miguel de Cervantes", 1605)
adicionar_livro("Orgulho e Preconceito", "Jane Austen", 1813)
adicionar_livro("1984", "George Orwell", 1949)
adicionar_livro("Cem Anos de Solidão", "Gabriel Garcia Marquez", 1967)
adicionar_livro("O Apanhador no Campo de Centeio", "J.D. Salinger", 1951)
adicionar_livro("O Segundo Sexo", "Simone de Beauvoir", 1949)


# Exibe os livros
listar_livros()


# Obtém os anos de publicação
anos = [
    livro.ano_publicacao
    for livro in biblioteca
]


# Remove anos repetidos e ordena
anos = list(set(anos))
anos.sort()


# Conta os livros de cada ano
contagem_por_ano = [
    anos.count(ano)
    for ano in anos
]


# Exibe a quantidade por ano
print("\n===== QUANTIDADE POR ANO =====")

for i, ano in enumerate(anos):
    print(f"Ano {ano}: {contagem_por_ano[i]} livro(s)")


# Cria o gráfico
plt.plot(
    anos,
    contagem_por_ano,
    marker="o",
    linestyle="-"
)


# Adiciona rótulos aos eixos
plt.xlabel("Ano de Publicação")
plt.ylabel("Número de Livros")

# Adiciona título
plt.title("Distribuição de Livros por Ano")


# Mostra os valores nos pontos
for i, valor in enumerate(contagem_por_ano):
    plt.text(
        anos[i],
        valor,
        str(valor),
        ha="center",
        va="bottom"
    )


# Adiciona uma grade
plt.grid(True)

# Exibe o gráfico
plt.show()



### Lógica resumida

#1. **Classe `Livro`** → define as características de cada livro.
#2. **Lista `biblioteca`** → armazena os livros.
#3. **`adicionar_livro()`** → cria e adiciona livros.
#4. **`listar_livros()`** → mostra os livros cadastrados.
#5. **`set()`** → remove anos repetidos.
#6. **`sort()`** → organiza os anos.
#7. **`count()`** → conta livros publicados em cada ano.
#8. **Matplotlib** → apresenta os resultados em um gráfico.


```
