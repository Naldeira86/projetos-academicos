```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
url = "https://raw.githubusercontent.com/wcota/covid19br/master/cases-brazil-states.csv"
df = pd.read_csv(url)
df_raw = pd.read_csv(url)
df.head()
```

```python

df = df_raw[df_raw['state'] == 'TOTAL'].copy()
```

```python
df.head()
```

```python
df['date'] = pd.to_datetime(df['date'])
df.head(1)
```

```python
df = df[['date', 'newCases', 'newDeaths', 'totalCases']].reset_index(drop=True)
```

```python
df.head(2)
```

```python
print(f"Dados de {df['date'].min().date()} até {df['date'].max().date()}")
print(f"Total de registros: {len(df)}")
```

```python
# Cena 2 - " A curcva completa " (3 min)
# Pergunta pro aluno antes de rodar: "Vocês lembram quando foi pico?"
fig, ax = plt.subplots(figsize=(13, 5))

ax.plot(df['date'], df['newCases'], color='steelblue', linewidth=1.5)

ax.fill_between(df['date'], df['newCases'], alpha=0.25, color='steelblue')

ax.set_title("Novos casos por dia - Brasil (pandemia completa)", fontsize=14)

ax.set_xlabel("Data")

ax.set_ylabel("Novos casos")

plt.tight_layout()

plt.show()


```

```python
ondas = [
    ("1ª onda ", "2020-07-15", "steelblue"),
    ("2ª onda ", "2021-03-25", "tomato"),
    ("3ª onda ", "2022-02-01", "steelblue")
]
```

```python
ondas
```

```python
# Cena 3 - "Casos vs Mortes - a história muda" (4 min)
fig, ax = plt.subplots(figsize=(13, 5))

ax.plot(df['date'], df['newCases'], color='gray', linewidth=1.2)

ax.fill_between(df['date'], df['newCases'], alpha=0.15, color='gray')

for nome, data, cor in ondas:
    ax.axvline(pd.to_datetime(data), color=cor, linestyle='--', linewidth=1.5, label=nome)

ax.legend()

ax.set_title("Novos casos por dia - Brasil (pandemia completa)", fontsize=14)

ax.set_xlabel("Data")

ax.set_ylabel("Novos casos")

plt.tight_layout()
plt.show()

```

```python

# Cena 3 - "Casos vs Mortes - a história muda" (4 min)

fig, ax1 = plt.subplots(figsize=(13, 5))

ax1.plot(
    df['date'],
    df['newCases'],
    color='steelblue',
    label='Novos casos'
)

ax1.set_ylabel("Novos casos", color='steelblue')

ax2 = ax1.twinx()

ax2.plot(
    df['date'],
    df['newDeaths'],
    color='tomato',
    label='Mortes',
    linewidth=1.5
)

ax2.set_ylabel("Novas mortes", color='tomato')

ax1.set_title(
    "Casos vs Mortes - por que a curva de mortes não acompanha os casos?",
    fontsize=14
)

fig.legend(
    loc='upper left',
    bbox_to_anchor=(0.01, 0.88)
)

plt.tight_layout()
plt.show()

```

```python
# Cria a coluna "mes" usando a data no formato ano-mês
df['mes'] = df['date'].dt.to_period('M').astype(str)

# Define o tamanho do gráfico
plt.figure(figsize=(14, 5))

# Cria o gráfico de barras somando as mortes de cada mês
sns.barplot(
    data=df,
    x='mes',
    y='newDeaths',
    estimator='sum',
    errorbar=None,
    color='tomato'
)

# Define o título do gráfico
plt.title("Total de mortes por mês - Brasil", fontsize=14)

# Define o nome do eixo X
plt.xlabel("Mês")

# Define o nome do eixo Y
plt.ylabel("Mortes no mês")

# Gira os nomes dos meses para facilitar a leitura
plt.xticks(rotation=45)

# Ajusta o espaço do gráfico
plt.tight_layout()

# Exibe o gráfico
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
df = sns.load_dataset('tips')
plt.figure(figsize=(8, 5))
sns.barplot(x='time', y='total_bill', data=df, estimator=sum, ci=None, palette="Set2")
plt.xlabel('Período (Time)')

plt.ylabel('Total de Gastos')

plt.title('Total de Gastos por Período (Almoço ou Jantar)')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(x='time', y='total_bill', data=df)#, #estimator=sum, ci=None, palette=“Set2”)

plt.xlabel('Período (Time)')

plt.ylabel('Média de Gastos')

plt.title('Média de Gastos por Período (Almoço ou Jantar)')

plt.show()
```

```python
# Crie um gráfico de barras com o Seaborn para mostrar a média de gorjetas por período

plt.figure(figsize=(8, 5))

sns.barplot(x='time', y='total_bill', data=df, palette="Set3")

plt.xlabel('Período (Time)')

plt.ylabel('Média da Gorjeta')

plt.title('Média da Gorjeta por Período (Almoço ou Jantar)')

plt.show()
```
