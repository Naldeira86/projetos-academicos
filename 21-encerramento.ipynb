## ENCERRAMENTO

```python

# ============================================================
# FUNÇÃO PARA CALCULAR O DESCONTO
# ============================================================
# Esta função recebe o valor do produto e o percentual de desconto.
def calcular_desconto(valor, percentual):

    # Verifica se o percentual está fora do intervalo permitido.
    # O desconto deve ser no mínimo 0% e no máximo 100%.
    if percentual < 0 or percentual > 100:

        # Retorna None para informar que o desconto é inválido.
        return None

    # Calcula quanto será descontado do valor original.
    # Exemplo: R$ 100 com 10% de desconto = R$ 10 de desconto.
    desconto = valor * (percentual / 100)

    # Retorna o valor final do produto depois de aplicar o desconto.
    return valor - desconto


# ============================================================
# FUNÇÃO PARA REGISTRAR A VENDA
# ============================================================
# Esta função recebe o nome do produto e o valor final da venda.
def registrar_venda(produto, valor_final):

    # Mostra na tela que a venda foi registrada.
    # :.2f garante que o valor apareça com duas casas decimais.
    print(f"Venda registrada: {produto} por R$ {valor_final:.2f}")


# ============================================================
# FUNÇÃO PARA ARREDONDAR VALORES
# ============================================================
# Lambda é uma forma curta de criar uma função.
# Neste caso, ela recebe um valor e arredonda para 2 casas decimais.
arredondar = lambda v: round(v, 2)


# ============================================================
# VARIÁVEIS DO CAIXA
# ============================================================
# Guarda o valor total de todas as vendas realizadas.
# Começamos com 0 porque nenhuma venda foi realizada ainda.
total_vendas = 0

# Guarda a quantidade de vendas realizadas.
# Também começa em 0.
qtd_vendas = 0


# ============================================================
# LOOP PRINCIPAL DO PDV
# ============================================================
# while True cria um loop que continuará executando
# até que exista uma condição para interrompê-lo.
while True:

    # Exibe o título do sistema.
    print("\n--- MINHA LOJA PDV ---")

    # Exibe as opções disponíveis para o usuário.
    print("1. Registrar venda")
    print("2. Ver relatório do dia")
    print("3. Encerrar caixa")

    # Pede ao usuário que escolha uma opção.
    # input() sempre retorna um texto (string).
    opcao = input("\nEscolha uma opção: ")


    # ========================================================
    # OPÇÃO 1 — REGISTRAR VENDA
    # ========================================================
    # Verifica se o usuário escolheu a opção 1.
    if opcao == "1":

        # Solicita o nome do produto.
        produto = input("Nome do produto: ")

        # Solicita o valor do produto.
        # float() transforma o texto digitado em número decimal.
        valor = float(input("Valor do produto: R$ "))

        # Solicita o percentual de desconto.
        # float() permite trabalhar com valores como 10.5%.
        percentual = float(input("Desconto (%): "))


        # Envia o valor e o percentual para a função
        # calcular_desconto().
        # A função calcula e devolve o valor final.
        valor_final = calcular_desconto(valor, percentual)


        # Verifica se a função encontrou um desconto inválido.
        # Quando o desconto é inválido, a função retorna None.
        if valor_final is None:

            # Informa ao usuário que o desconto não é permitido.
            print("Desconto inválido. Use um valor entre 0 e 100.")


        # Caso o desconto seja válido, executamos esta parte.
        else:

            # Arredonda o valor final para duas casas decimais.
            valor_final = arredondar(valor_final)

            # Chama a função registrar_venda()
            # para mostrar a venda registrada.
            registrar_venda(produto, valor_final)

            # Soma o valor da venda ao total do caixa.
            total_vendas = total_vendas + valor_final

            # Aumenta a quantidade de vendas realizadas em 1.
            qtd_vendas = qtd_vendas + 1


    # ========================================================
    # OPÇÃO 2 — RELATÓRIO DO DIA
    # ========================================================
    # Verifica se o usuário escolheu a opção 2.
    elif opcao == "2":

        # Mostra o título do relatório.
        print("\n--- RELATÓRIO DO DIA ---")

        # Mostra a quantidade de vendas realizadas.
        print(f"Quantidade de vendas: {qtd_vendas}")

        # Mostra o valor total faturado.
        # A função arredondar() garante duas casas decimais.
        print(f"Total faturado: R$ {arredondar(total_vendas):.2f}")


    # ========================================================
    # OPÇÃO 3 — ENCERRAR CAIXA
    # ========================================================
    # Verifica se o usuário escolheu a opção 3.
    elif opcao == "3":

        # Informa que o caixa está sendo encerrado.
        print("\nEncerrando caixa...")

        # Mostra a quantidade total de vendas realizadas.
        print(f"Vendas realizadas: {qtd_vendas}")

        # Mostra o valor total faturado durante o funcionamento
        # do caixa.
        print(f"Total faturado: R$ {arredondar(total_vendas):.2f}")
        break


    # ========================================================
    # OPÇÃO INVÁLIDA
    # ========================================================
    # Se o usuário digitar qualquer coisa diferente de
    # 1, 2 ou 3, nenhuma das condições anteriores será atendida.
    # Então o programa chega ao else.
    else:

        # Informa que a opção escolhida não existe.
        print("Opção inválida. Tente novamente.")

```
