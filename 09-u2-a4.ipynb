```python
# 1. A primeira forma - import math (usa math.função)
import math
print(math.sqrt(25)) # Raiz quadrada
print(math.log2(1024)) # logaritimo na base 2
print(math.cos(45)) # cosseno
```

```python
# 2. A segunda forma - impor math as m ou s
import math as s
print(s.sqrt(25))   # Raiz quadrada
print(s.log2(1024)) # logaritimo na base 2
print(s.cos(45))    # cosseno
```

```python
# 3. A terceira forma - from math import (carrega só o que precisa)
from math import sqrt, log2, cos
print(sqrt(25))
print(log2(1024))
print(cos(45))
```

```python
# 4. Matplotlib - primeiro gráfico de linha
import matplotlib.pyplot as plt

# Dados
x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 3, 5]

# Criar um gráfico de linha
plt.plot(x, y)

# Adicionar rótulos aos eixos
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')

# Adicionar um título
plt.title('Meu primeiro gráfico')

# Mostrar o gráfico
plt.show()
```

```python
# 5. Vamos exercitar - gráfico de vendas
import matplotlib.pyplot as plt

# Dados
meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai']
vendas = [120, 90, 150, 80, 200]

# criar um gráfico de barras
plt.bar(meses, vendas, color='royalblue')
# rotulos e titulos
plt.xlabel('Meses')
plt.ylabel('Vendas')
plt.title('Vendas Mensais')
# mostrar o grafico
plt.show()
```
