```python

# Cria o arquivo HTML
html = '''
<!DOCTYPE html>
<html lang="pt-BR">

<head>

    <!-- Define a codificação dos caracteres -->
    <meta charset="UTF-8">

    <!-- Faz a página se adaptar ao celular -->
    <meta name="viewport" content="width=device-width, initial-scale=1.0">

    <!-- Título da página -->
    <title>Tabs - Kivy convertido para HTML</title>

    <style>

        /* Configura a página */
        body {
            margin: 0;
            padding: 0;

            font-family: Arial, sans-serif;

            background-color: #f2f2f2;
        }


        /* Caixa principal */
        .tabs-container {
            width: 80%;
            max-width: 700px;

            margin: 50px auto;

            background-color: white;

            border-radius: 10px;

            box-shadow: 0 4px 10px rgba(0, 0, 0, 0.2);

            overflow: hidden;
        }


        /* Área dos botões */
        .tabs {
            display: flex;

            background-color: #333;
        }


        /* Botões das abas */
        .tab-button {
            flex: 1;

            padding: 15px;

            border: none;

            background-color: #333;

            color: white;

            font-size: 16px;

            cursor: pointer;
        }


        /* Efeito quando passa o mouse */
        .tab-button:hover {
            background-color: #555;
        }


        /* Botão selecionado */
        .tab-button.active {
            background-color: #2196F3;
        }


        /* Conteúdo das abas */
        .tab-content {
            display: none;

            padding: 30px;

            text-align: center;
        }


        /* Mostra a aba selecionada */
        .tab-content.active {
            display: block;
        }


        /* Título */
        .tab-content h2 {
            margin-bottom: 15px;
        }


        /* Texto */
        .tab-content p {
            font-size: 18px;
        }

    </style>

</head>


<body>


    <!-- Caixa principal -->
    <div class="tabs-container">


        <!-- Botões das abas -->
        <div class="tabs">

            <!-- Aba 1 -->
            <button
                class="tab-button active"
                onclick="mostrarAba('tab1', this)">
                Tab 1
            </button>


            <!-- Aba 2 -->
            <button
                class="tab-button"
                onclick="mostrarAba('tab2', this)">
                Tab 2
            </button>

        </div>


        <!-- Conteúdo da Tab 1 -->
        <div id="tab1" class="tab-content active">

            <h2>Tab 1</h2>

            <p>Content for Tab 1</p>

        </div>


        <!-- Conteúdo da Tab 2 -->
        <div id="tab2" class="tab-content">

            <h2>Tab 2</h2>

            <p>Content for Tab 2</p>

        </div>


    </div>


    <script>

        // Função para trocar de aba
        function mostrarAba(nomeAba, botao) {

            // Seleciona todos os conteúdos
            const conteudos =
                document.querySelectorAll('.tab-content');


            // Esconde todos os conteúdos
            conteudos.forEach(function(conteudo) {

                conteudo.classList.remove('active');

            });


            // Seleciona todos os botões
            const botoes =
                document.querySelectorAll('.tab-button');


            // Remove o destaque dos botões
            botoes.forEach(function(botaoAtual) {

                botaoAtual.classList.remove('active');

            });


            // Mostra a aba selecionada
            document
                .getElementById(nomeAba)
                .classList.add('active');


            // Destaca o botão selecionado
            botao.classList.add('active');

        }

    </script>


</body>

</html>
'''


# Salva o código em um arquivo HTML
with open("tabs.html", "w", encoding="utf-8") as arquivo:

    arquivo.write(html)


# Exibe o HTML dentro do Google Colab
from IPython.display import HTML, display

display(HTML(html))

```

```python

# Código da calculadora em HTML, CSS e JavaScript

html = '''
<!DOCTYPE html>

<html lang="pt-BR">

<head>

    <!-- Define a codificação dos caracteres -->
    <meta charset="UTF-8">

    <!-- Adapta a página para celulares -->
    <meta name="viewport" content="width=device-width, initial-scale=1.0">

    <!-- Título da página -->
    <title>Calculadora</title>


    <style>

        /* Configuração da página */
        body {

            margin: 0;

            padding: 0;

            font-family: Arial, sans-serif;

            background-color: #f2f2f2;

        }


        /* Caixa da calculadora */
        .calculadora {

            width: 350px;

            margin: 50px auto;

            padding: 20px;

            background-color: white;

            border-radius: 15px;

            box-shadow: 0 5px 15px rgba(0, 0, 0, 0.2);

        }


        /* Título */
        h1 {

            text-align: center;

            margin-bottom: 20px;

        }


        /* Campo de entrada */
        #display {

            width: 100%;

            box-sizing: border-box;

            padding: 15px;

            margin-bottom: 15px;

            font-size: 22px;

            text-align: right;

            border: 1px solid #ccc;

            border-radius: 8px;

        }


        /* Grade dos botões */
        .botoes {

            display: grid;

            grid-template-columns: repeat(4, 1fr);

            gap: 10px;

        }


        /* Botões */
        button {

            padding: 20px;

            font-size: 20px;

            border: none;

            border-radius: 8px;

            background-color: #2196F3;

            color: white;

            cursor: pointer;

        }


        /* Efeito do botão */
        button:hover {

            background-color: #1976D2;

        }


        /* Botões das operações */
        .operador {

            background-color: #FF9800;

        }


        /* Efeito dos operadores */
        .operador:hover {

            background-color: #F57C00;

        }


        /* Botão limpar */
        .limpar {

            background-color: #F44336;

        }


        /* Efeito do botão limpar */
        .limpar:hover {

            background-color: #D32F2F;

        }


        /* Botão igual */
        .igual {

            background-color: #4CAF50;

        }


        /* Efeito do botão igual */
        .igual:hover {

            background-color: #388E3C;

        }

    </style>

</head>


<body>


    <!-- Calculadora -->
    <div class="calculadora">


        <!-- Título -->
        <h1>Calculadora</h1>


        <!-- Campo onde aparece a conta -->
        <input
            type="text"
            id="display"
            placeholder="Insira um número"
            readonly
        >


        <!-- Botões -->
        <div class="botoes">


            <!-- Número 1 -->
            <button onclick="adicionarNumero('1')">
                1
            </button>


            <!-- Número 2 -->
            <button onclick="adicionarNumero('2')">
                2
            </button>


            <!-- Número 3 -->
            <button onclick="adicionarNumero('3')">
                3
            </button>


            <!-- Soma -->
            <button
                class="operador"
                onclick="adicionarOperador('+')">
                +
            </button>


            <!-- Número 4 -->
            <button onclick="adicionarNumero('4')">
                4
            </button>


            <!-- Número 5 -->
            <button onclick="adicionarNumero('5')">
                5
            </button>


            <!-- Número 6 -->
            <button onclick="adicionarNumero('6')">
                6
            </button>


            <!-- Subtração -->
            <button
                class="operador"
                onclick="adicionarOperador('-')">
                -
            </button>


            <!-- Número 7 -->
            <button onclick="adicionarNumero('7')">
                7
            </button>


            <!-- Número 8 -->
            <button onclick="adicionarNumero('8')">
                8
            </button>


            <!-- Número 9 -->
            <button onclick="adicionarNumero('9')">
                9
            </button>


            <!-- Multiplicação -->
            <button
                class="operador"
                onclick="adicionarOperador('*')">
                *
            </button>


            <!-- Limpar -->
            <button
                class="limpar"
                onclick="limpar()">
                C
            </button>


            <!-- Número 0 -->
            <button onclick="adicionarNumero('0')">
                0
            </button>


            <!-- Resultado -->
            <button
                class="igual"
                onclick="calcular()">
                =
            </button>


            <!-- Divisão -->
            <button
                class="operador"
                onclick="adicionarOperador('/')">
                /
            </button>


        </div>

    </div>


    <script>


        // Pega o campo da calculadora
        const display = document.getElementById("display");


        // Adiciona um número no campo
        function adicionarNumero(numero) {

            display.value = display.value + numero;

        }


        // Adiciona uma operação
        function adicionarOperador(operador) {

            display.value = display.value + operador;

        }


        // Limpa o campo
        function limpar() {

            display.value = "";

        }


        // Calcula o resultado
        function calcular() {

            try {

                // Calcula a expressão
                let resultado = eval(display.value);

                // Mostra o resultado
                display.value = resultado;

            }

            catch (erro) {

                // Mostra erro caso a conta seja inválida
                display.value = "Erro";

            }

        }


    </script>


</body>

</html>
'''


# Salva a calculadora em um arquivo HTML
with open("calculadora.html", "w", encoding="utf-8") as arquivo:

    arquivo.write(html)


# Importa recursos para mostrar HTML no Colab
from IPython.display import HTML, display


# Mostra a calculadora no notebook
display(HTML(html))

```
