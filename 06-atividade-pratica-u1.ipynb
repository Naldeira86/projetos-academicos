```python

# GESTAO_DE_NOTAS_DE_ALUNO

def calcular_media(notas):             # Cria a função para calcular a média
    total = sum(notas)                 # Soma todas as notas
    media = total / len(notas)         # Divide a soma pela quantidade de notas
    return media                       # Retorna a média

notas = []  # Cria uma lista vazia para guardar as notas

# Pergunta quantas notas serão digitadas
quantidade = int(input("Quantas notas deseja inserir? "))

for i in range(quantidade): # Repete o código pela quantidade informada
    nota = float(input("Digite a nota: ")) # Recebe uma nota do usuário
    notas.append(nota) # Adiciona a nota na lista

media = calcular_media(notas) # Chama a função e calcula a média

# Cria função para arredondar para 2 casas
arredondar_media = lambda media: round(media, 2)
media = arredondar_media(media) # Arredonda a média calculada

if media >= 7:                 # Verifica se a média é maior ou igual a 7
    situacao = "Aprovado"      # Se for 7 ou mais, aluno está aprovado
else:                          # Caso contrário
    situacao = "Reprovado"     # Aluno está reprovado

print("\n===== RELATÓRIO FINAL =====")  # Exibe o título do relatório
print("Notas inseridas:", notas)        # Exibe as notas digitadas
print("Média:", media)                  # Exibe a média
print("Situação:", situacao)            # Exibe aprovado ou reprovado

```
