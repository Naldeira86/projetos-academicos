```python
exemplo2 = {'a': 10, 'b': 20, 'c': 30, 'd': 40, 'e': 50}
series2 = pd.Series(data=exemplo2)
print(series2)
```

```python
# Metodos prontos da Series.
print("Média:", series2.mean())
print("Soma:", series2.sum())
print("Máximo:", series2.max())
print("Mínimo:", series2.min())
```

```python
# Idade média dos clientes
import pandas as pd

# Dados dos clientes
dados = {
    'Nome': ['João', 'Maria', 'Pedro', 'Ana', 'Carlos'],
    'Idade': [25, 30, 35, 28, 40]
}

# Criar uma Series com idades indexadas pelo nome
series_idades = pd.Series(data=dados['Idade'], index=dados['Nome'])

print("Series de idades:")
print(series_idades)

# Calcular a idade média
idade_media = series_idades.mean()

print("Idade média dos clientes:", idade_media)

# Outras estatísticas que ajudam na decisão de Marketing
print(f"Idade mínima dos clientes: {series_idades.min()}")
print(f"Idade máxima dos clientes: {series_idades.max()}")
```
