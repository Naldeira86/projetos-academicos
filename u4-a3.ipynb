```python
def calcular_total(valores):
    """
    Calcula o total de uma lista de valores de compra.

    Guarda 1:
    A lista não pode ser None.

    Guarda 2:
    Nenhum valor pode ser negativo.
    """

    assert valores is not None, "A lista não pode ser None"
    assert all(v >= 0 for v in valores), "Não são permitidos valores negativos"

    return sum(valores)


compra = [29.90, 15.00, 5.50]

print(f"Total da compra: R$ {calcular_total(compra):.2f}")

print("Teste 1:", calcular_total([29.90, 15.00, 5.50]))

print("Teste 2:", calcular_total([]))

assert calcular_total([10, 20, 30]) == 60, "Total incorreto"

print("Teste 3: assert passou - resultado correto!")


def aplicar_desconto(valor, percentual):
    """
    Aplica um desconto percentual sobre um valor.

    Parâmetros:
        valor (float): valor original da compra.
        percentual (float): percentual de desconto.

    Retorna:
        float: valor final depois do desconto.
    """

    assert valor >= 0, "O valor não pode ser negativo"
    assert 0 <= percentual <= 100, \
        "O percentual deve estar entre 0 e 100"

    desconto = valor * percentual / 100
    valor_final = valor - desconto

    return valor_final


print(f"Desconto: R$ {aplicar_desconto(100, 10):.2f}")

print(f"Desconto: R$ {aplicar_desconto(200, 20):.2f}")

print(f"Sem desconto: R$ {aplicar_desconto(100, 0):.2f}")

assert aplicar_desconto(100, 10) == 90

print("Teste de desconto: assert passou!")


def calcular_total_doctest(valores):
    """
    Calcula o total de uma lista de valores de compra.

    Exemplos:
    >>> calcular_total_doctest([10, 20, 30])
    60

    >>> calcular_total_doctest([20.90, 15.00, 5.50])
    41.4

    >>> calcular_total_doctest([])
    0
    """

    return sum(valores)


import doctest

resultado = doctest.testmod(verbose=True)

print(
    f"\n{resultado.attempted} testes rodados, "
    f"{resultado.failed} falharam"
)


import unittest


class TestCalcularTotal(unittest.TestCase):

    def test_compra_normal(self):

        self.assertEqual(
            calcular_total([10, 20, 30]),
            60
        )

    def test_lista_vazia(self):

        self.assertEqual(
            calcular_total([]),
            0
        )

    def test_valores_decimais(self):

        self.assertAlmostEqual(
            calcular_total([20.90, 15.00, 5.50]),
            41.40,
            places=2
        )

    def test_um_item(self):

        self.assertEqual(
            calcular_total([99]),
            99
        )

    def test_aplicar_desconto(self):

        self.assertEqual(
            aplicar_desconto(100, 10),
            90
        )

    def test_desconto_vinte_porcento(self):

        self.assertEqual(
            aplicar_desconto(200, 20),
            160
        )

    def test_sem_desconto(self):

        self.assertEqual(
            aplicar_desconto(100, 0),
            100
        )

    def test_desconto_total(self):

        self.assertEqual(
            aplicar_desconto(100, 100),
            0
        )


unittest.main(
    argv=[''],
    exit=False,
    verbosity=2
)
```
