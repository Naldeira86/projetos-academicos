```python


import sqlite3

conn = sqlite3.connect('Minha_loja.db')

cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS Vendas (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    data TEXT,
    produto TEXT,
    categoria TEXT,
    quantidade INTEGER,
    valor_unitario REAL
    )
""")
cursor.execute("DELETE FROM Vendas")
conn.commit()
print("Banco da minha loja pronto!")

```

```python
# Inserir vendas
vendas_do_ano = [
    ('2024-01-15', 'Notebook',   'Informática',     1, 3500.00, ),
    ('2024-02-20', 'Smartphone', 'Eletrônicos',     2, 1500.00,),
    ('2024-03-10', 'Camiseta',   'Vestuário',       5, 25.00, ),
    ('2024-04-05', 'Livro',      'Livros',          1, 45.00, ),
    ('2024-05-12', 'Tênis',      'Calçados',        3, 80.00, ),
    ('2024-06-25', 'Tablet',     'Informática',     2, 2000.00, ),
    ('2024-07-18', 'Câmera',     'Eletrônicos',     1, 500.00,),
    ('2024-08-30', 'Bermuda',    'Vestuário',       4, 30.00, ),
    ('2024-09-14', 'Garrafa',    'Bebidas',         6, 10.00, ),
    ('2024-10-08', 'Mouse',      'Informática',     3, 20.00, ),
    ('2024-11-22', 'Fone de Ouvido', 'Eletrônicos', 2, 50.00, ),
    ('2024-12-15', 'Calça',      'Vestuário',       3, 40.00, ),
    ('2025-01-01', 'Caderno',    'Livros',          2, 15.00, ),
    ('2025-02-15', 'Sapato',     'Calçados',        2, 70.00, ),
    ('2025-03-20', 'Monitor',    'Informática',     1, 800.00, ),
    ('2025-04-10', 'Garrafa',    'Bebidas',         4, 10.00, ),
]
cursor.executemany("INSERT INTO Vendas (data, produto, categoria, quantidade, valor_unitario) VALUES (?, ?, ?, ?, ?)", vendas_do_ano)
conn.commit()
print(f"{len(vendas_do_ano)} Vendas inseridas com sucesso!")
```

```python
# LER DIRETO PARA O DATAFRAME + FECHAR CONEXÃO.
import pandas as pd
df = pd.read_sql("SELECT * FROM Vendas", conn)
conn.close()
df['total'] = df['quantidade'] * df['valor_unitario']
df['data'] = pd.to_datetime(df['data'])
df['mes'] = df['data'].dt.month
df['mes_nome'] = df['data'].dt.strftime('%b') # JAN, FEV, MAR, ...

print(f"Faturamento total da minha loja: R$ {df['total'].sum():,.2f}" )
df.head()

```

```python
# 4 - Filtro booleano: destaque para os produtos vendidos acima de R$ 1.000

# Filtra somente as vendas com total acima de R$ 1.000
acima_de_1000 = df[df['total'] > 1000]

# Mostra a quantidade de vendas acima de R$ 1.000
print(f"Vendas acima de R$ 1.000,00: {len(acima_de_1000)} de {len(df)} registros\n")

# Seleciona as colunas e ordena pelo total, do maior para o menor
print(
    acima_de_1000[['data', 'produto', 'quantidade', 'total']]
    .sort_values(by='total', ascending=False)
    .to_string(index=False)
)

```
