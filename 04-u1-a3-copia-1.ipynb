```python
filme = ['Filme 1', 'Filme 2', 'Filme 3', 'Filme 4', 'Filme 5']
print("Bem vindo a classificação do filme!")
print("Você tem cinco filmes para classificar!")
print("Digite '0' para parar...")

for filmes in filme:
      classificacao = input(f"Como você classifica '{filmes}' de 1 á 5 (ou '0' para parar:) ")
      if classificacao == '0':
        print("Que pena!")
        break
      classificacao = int(classificacao)
      if classificacao <1 or classificacao >5:
        print("Nota invalida!!!")
      else:
        print(f"Você classificou {filmes} com a {classificacao} estrelas: \n")
        print("Tank you!")
```
