```python
def e_impar(numero):
    return numero % 2 != 0


# Testando a função
numero = 123120

if e_impar(numero):
    print(f"{numero} é um número ímpar.")
else:
    print(f"{numero} não é um número ímpar.")


# Testando a função
numero = 1355989

if e_impar(numero):
    print(f"{numero} é um número ímpar.")
else:
    print(f"{numero} não é um número ímpar.")
```

```python
# Função para calcular a média
def calcular_media(notas):
    total = sum(notas)
    media = total / len(notas)
    return media


# Função para verificar a situação do aluno
def media_aluno(media):
    if media >= 7:
        return "Aprovado"
    else:
        return "Reprovado"


# Função lambda para arredondar a média
arredondar_media = lambda media: round(media, 2)


# Entrada das notas
N1 = float(input("Digite a nota 1: "))
N2 = float(input("Digite a nota 2: "))


# Criando a lista de notas
notas = [N1, N2]


# Calculando a média
media = calcular_media(notas)


# Arredondando a média
media_arredondada = arredondar_media(media)


# Verificando a situação
situacao = media_aluno(media_arredondada)


# Resultados
print("\nNotas:", notas)
print("A média do aluno foi:", media_arredondada)
print("Situação:", situacao)
```
