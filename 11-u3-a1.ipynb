```python
import sqlite3
conn = sqlite3.connect('Minha_loja.db')
cursor = conn.cursor()


# CREATE
cursor.execute("""
CREATE TABLE IF NOT EXISTS Produtos (
    id INTEGER PRIMARY KEY,
    nome TEXT NOT NULL,
    preco REAL NOT NULL,
    estoque INTEGER
)
""")

# INSERT
cursor.execute("""
INSERT INTO Produtos (nome, preco, estoque)
VALUES (?, ?, ?)
""", ("Camiseta", 19.99, 50))

# SELECT
cursor.execute("SELECT * FROM Produtos")

produtos = cursor.fetchall()

for produto in produtos:
    print(produto)

# UPDATE
cursor.execute(
    "UPDATE Produtos SET preco = ? WHERE id = ?",
    (27.99, 1)
)

# DELETE
cursor.execute(
    "DELETE FROM Produtos WHERE id = ?",
    (1,)
)

# Salva as alterações
conn.commit()

# Fecha a conexão
conn.close()
```
